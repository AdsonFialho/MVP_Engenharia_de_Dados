```python
%sql
-- Cria o catálogo do projeto de e-commerce brasileiro.
-- Nomeado pelo domínio de negócio (não pela empresa Olist), permitindo
-- futuros catálogos por país (ecommerce_us, ecommerce_uk...) sem conflito.
CREATE CATALOG IF NOT EXISTS ecommerce_br;

-- Cria as três camadas da Arquitetura Medalhão como schemas dentro do catálogo.
-- Bronze: dados crus, exatamente como vieram da fonte.
-- Silver: dados limpos, padronizados e tipados.
-- Gold: modelo dimensional (esquema estrela) pronto para análise.
CREATE SCHEMA IF NOT EXISTS ecommerce_br.bronze;
CREATE SCHEMA IF NOT EXISTS ecommerce_br.silver;
CREATE SCHEMA IF NOT EXISTS ecommerce_br.gold;
```

```python
%sql
-- Confirma que o catálogo e os três schemas foram criados corretamente.
SHOW SCHEMAS IN ecommerce_br;
```

```python
%sql
-- Cria um Volume na camada Bronze para armazenar os arquivos CSV brutos do Olist.
-- O Volume é o "storage" do Lakehouse: guarda os arquivos como vieram da fonte,
-- antes de serem lidos e convertidos em tabelas Delta.
CREATE VOLUME IF NOT EXISTS ecommerce_br.bronze.raw_files;
```
